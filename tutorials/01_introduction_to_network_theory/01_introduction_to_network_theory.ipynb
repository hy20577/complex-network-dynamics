## The Birth of Graph Theory

The history of graph theory in mathematical problem solving dates back to 1736 [[2](#ref-2)]. Swiss-born
mathematician Leonhard Euler, who was working in St. Petersburg at the time, encountered a
question that was famous among the people of the Prussian city of Königsberg (now Kaliningrad):
could one walk through the city crossing each of its bridges exactly once? The river Pregel
divides the city into four land masses, which were connected by seven bridges (Figure 1).

```python
from urllib.request import urlopen

from PIL import Image

picture = Image.open(urlopen("https://hy20577.github.io/complex-network-dynamics/tutorials/images/bridge.jpg"))

# The bridges and the two land masses each of them joins
bridges = {"a": "AB", "b": "AB", "c": "AC", "d": "AC", "e": "AD", "f": "BD", "g": "CD"}
position = {"A": (0.0, 0.0), "B": (0.0, -1.0), "C": (0.0, 1.0), "D": (1.6, 0.0)}
bend = {"a": 0.3, "b": -0.3, "c": -0.3, "d": 0.3, "e": 0.0, "f": 0.0, "g": 0.0}  # two bridges join A-B and A-C

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5), gridspec_kw={"width_ratios": [1.5, 1]})

ax1.imshow(picture, cmap="gray")
ax1.set_title("(a) The seven bridges")
ax1.axis("off")

for bridge, (u, v) in bridges.items():
    (x1, y1), (x2, y2) = position[u], position[v]
    ax2.annotate("", xy=(x2, y2), xytext=(x1, y1),
                 arrowprops=dict(arrowstyle="-", color="gray", linewidth=2, connectionstyle=f"arc3,rad={bend[bridge]}"))
    # the label sits on the middle of the (curved) link
    ax2.text((x1 + x2) / 2 + 0.5 * bend[bridge] * (y2 - y1), (y1 + y2) / 2 - 0.5 * bend[bridge] * (x2 - x1), bridge,
             ha="center", va="center", fontsize=12, style="italic", bbox=dict(boxstyle="circle", fc="white", ec="none"))
for node, (x, y) in position.items():
    ax2.scatter(x, y, s=900, color="skyblue", edgecolor="black", zorder=3)
    ax2.text(x, y, node, ha="center", va="center", fontsize=13, fontweight="bold", zorder=4)
ax2.set_xlim(-0.6, 2.0)
ax2.set_ylim(-1.3, 1.3)
ax2.set_aspect("equal")
ax2.set_title("(b) The corresponding graph")
ax2.axis("off")
plt.tight_layout()
plt.show()
```

**Figure 1:** (a) The seven bridges of Königsberg, where uppercase letters denote the land masses
and lowercase letters the bridges. (b) Its graphical representation, where the land masses
, B, C$ and $ are the vertices and the bridges the connecting lines, referred to as edges,
what is called a graph in mathematics.

Euler reformulated the question as a graphical problem. He assigned the uppercase letters
$A, B, C$ and $D$ to the land masses (referred to as vertices or nodes) and the lowercase letters
$a, b, c, d, e, f$ and $g$ to the bridges (referred to as edges or links). In 1736, he proved
that there is no way to cross every bridge exactly once.

Euler's solution was both simple and elegant: he counted the number of bridges connected to each
land mass, which we now call the degree of a node. For example, there are three bridges
connecting $B$ (or $C$) to the others, and the degrees of the nodes $A, B, C$ and $D$ are
$5, 3, 3$ and $3$, respectively. Whenever a walk enters a node through one bridge it has to leave
it through another, so every node must have an even degree, except possibly the two nodes where
the walk starts and ends. Hence, such a walk can exist only if zero or two nodes have an odd
degree. Here all four nodes have an odd degree, therefore, there is no way to cross all the
bridges exactly once.

This brings us to the mathematical definition of a graph [[3](#ref-3)].

> **Definition (Graph).** A graph $G$ is an ordered triple $(V(G), E(G), \psi_{G})$ consisting of
> a non-empty set $V(G)$ of vertices, a set $E(G)$, disjoint from $V(G)$, of edges, and an
> incidence function $\psi_{G}$ that associates with each edge of $G$ an unordered pair of (not
> necessarily distinct) vertices of $G$. If $e$ is an edge and $u$ and $v$ are vertices such that
> $\psi_{G}(e) = uv$, then $e$ is said to join $u$ and $v$; the vertices $u$ and $v$ are called
> the ends of $e$.

For the problem of the Königsberg bridges, the components of the graph are

$$
\begin{aligned}
G &= (V(G), E(G), \psi_{G}), \text{ where}\\
V(G) &= \{A, B, C, D\}, \\
E(G) &= \{a, b, c, d, e, f, g\},\\
\psi_{G}(a) &= AB,\ \psi_{G}(b) = AB,\ \psi_{G}(c) = AC,\ \psi_{G}(d) = AC,\\
\psi_{G}(e) &= AD,\ \psi_{G}(f) = BD,\ \psi_{G}(g) = CD.
\end{aligned}
$$

We can build this graph with NetworkX, from the list of bridges used for Figure 1(b), and repeat
Euler's argument. Two bridges join $A$ and $B$,
and two join $A$ and $C$, so we need a `MultiGraph`, which allows more than one link between the
same pair of nodes.

```python
konigsberg = nx.MultiGraph()
for bridge, (u, v) in bridges.items():
    konigsberg.add_edge(u, v, key=bridge)

degrees = dict(konigsberg.degree())
print("Degrees:", degrees)
print("Nodes with an odd degree:", sum(k % 2 for k in degrees.values()))
print("Is there a walk that crosses every bridge once?", nx.has_eulerian_path(konigsberg))
```

#### The 1875 Bridge

In 1875 a new bridge, which we call $h$, was built between the land masses $B$ and $C$. It adds 1
to the degrees of $B$ and $C$, so they become $5, 4, 4$ and $3$ for $A, B, C$ and $D$. Now only
two nodes, $A$ and $D$, have an odd degree, and, by Euler's argument, a walk that crosses every
bridge exactly once becomes possible: it has to start at one of the two odd nodes and end at the
other. We add the bridge to our graph, check it again, and let NetworkX find such a walk.

```python
konigsberg_1875 = konigsberg.copy()
konigsberg_1875.add_edge("B", "C", key="h")

degrees_1875 = dict(konigsberg_1875.degree())
print("Degrees:", degrees_1875)
print("Nodes with an odd degree:", [node for node, k in degrees_1875.items() if k % 2 == 1])
print("Is there a walk that crosses every bridge once?", nx.has_eulerian_path(konigsberg_1875))

walk = list(nx.eulerian_path(konigsberg_1875, keys=True))
print("\nOne such walk:")
print(walk[0][0] + "".join(f" --{bridge}--> {v}" for u, v, bridge in walk))
```

```python
def draw_bridges(ax, bridges, labels, colours, title):
    # Draw the land masses as nodes and the bridges as (curved) links, with a label on each link
    position = {"A": (0.0, 0.0), "B": (0.0, -1.0), "C": (0.0, 1.0), "D": (1.6, 0.0)}
    bend = {"a": 0.3, "b": -0.3, "c": -0.3, "d": 0.3, "e": 0.0, "f": 0.0, "g": 0.0, "h": -0.8}
    for bridge, (u, v) in bridges.items():
        (x1, y1), (x2, y2) = position[u], position[v]
        ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                    arrowprops=dict(arrowstyle="-", color=colours[bridge], linewidth=2.5,
                                    connectionstyle=f"arc3,rad={bend[bridge]}"))
        ax.text((x1 + x2) / 2 + 0.5 * bend[bridge] * (y2 - y1), (y1 + y2) / 2 - 0.5 * bend[bridge] * (x2 - x1),
                labels[bridge], ha="center", va="center", fontsize=11, style="italic",
                bbox=dict(boxstyle="round", fc="white", ec="none"))
    for node, (x, y) in position.items():
        ax.scatter(x, y, s=900, color="skyblue", edgecolor="black", zorder=3)
        ax.text(x, y, node, ha="center", va="center", fontsize=13, fontweight="bold", zorder=4)
    ax.set_xlim(-1.2, 2.0)
    ax.set_ylim(-1.3, 1.3)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.axis("off")


bridges_1875 = {**bridges, "h": "BC"}
order = {bridge: step for step, (u, v, bridge) in enumerate(walk, start=1)}

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
draw_bridges(ax1, bridges_1875, {b: b for b in bridges_1875},
             {b: ("crimson" if b == "h" else "gray") for b in bridges_1875},
             "(a) The bridges after 1875")
draw_bridges(ax2, bridges_1875, {b: f"{b}: {order[b]}" for b in bridges_1875},
             {b: plt.cm.viridis((order[b] - 1) / (len(order) - 1)) for b in bridges_1875},
             f"(b) A walk from {walk[0][0]} to {walk[-1][1]}")
plt.tight_layout()
plt.show()
```

**Figure 2:** (a) The graph of the Königsberg bridges after the new bridge $h$ (red) was built
between $B$ and $C$ in 1875. (b) A walk that crosses every bridge exactly once, found by NetworkX;
each bridge is labelled with the step at which it is crossed, and coloured from dark (first) to
yellow (last).

Note that the problem asks to cross every *bridge* once, not to visit every land mass once: the
walk above passes through $A$ several times, as it must, since five bridges meet there.

## Graph Theory to Network Science

The authors in [[4](#ref-4)] discussed the distinction between the terms "graph" and "network", which are commonly used interchangeably in the scientific literature. However, the difference lies in that a graph refers to the schematic visualisation used to solve a mathematical problem, whereas a network refers to the graphical representation of a real-life problem, such as transportation maps, social interactions, protein-protein interactions, etc.

Despite their close relationship, we shall henceforth refer to them as "networks" and to their components, vertices and edges, as "nodes" and "links", respectively.

## Regular Network

If all the $n$ nodes of a network have the same degree $d$, it is called a regular network (or
$d$-regular network). Every link adds 1 to the degree of each of its two ends, so the sum of all
degrees, $\sum_{i=1}^{n} d_{i} = n d$, is twice the number of links and must be an even number.
Moreover, $d \le n - 1$, since a node can be connected to at most $n - 1$ other nodes.

Below, we generate a random 3-regular network with 6 nodes ($n d = 18$, which is even).

```python
d = 3  # degree of every node
n = 6  # number of nodes (n * d = 18, which is even)

G = nx.random_regular_graph(d, n, seed=42)
G = nx.relabel_nodes(G, {node: node + 1 for node in G})  # number the nodes 1, 2, ..., n

plt.figure(figsize=(6, 6))
pos = nx.circular_layout(G)
nx.draw(G, pos, with_labels=True, node_color="skyblue", node_size=600, edge_color="gray", font_weight="bold")
plt.title(f"Random {d}-regular network with {n} nodes")
plt.show()
```

### The Adjacency Matrix

A network with $n$ nodes can be stored in an $n \times n$ adjacency matrix $A$, where $A_{ij} = 1$
if nodes $i$ and $j$ are linked and $A_{ij} = 0$ otherwise. For an undirected network $A$ is
symmetric, and the sum of row $i$ is the degree of node $i$, $d_{i} = \sum_{j} A_{ij}$. Hence,
the sum of all the entries of $A$ equals $n d = 18$ for our network.

```python
A = nx.to_numpy_array(G, dtype=int)
print(A)
print("Degree of each node:", A.sum(axis=1))
print("Sum of all degrees: ", A.sum())
```

### Degree Distribution

The degree distribution shows how many nodes have each degree. In a regular network there is
only one column, as the degrees of all nodes are equal. The network types of the
[next tutorial](02_network_types.ipynb) differ mainly in the shape of this distribution.

```python
degrees = [k for node, k in G.degree()]

plt.figure(figsize=(7, 4))
plt.hist(degrees, bins=range(min(degrees) - 2, max(degrees) + 2), align="left", rwidth=0.6,
         color="steelblue", edgecolor="black")
plt.title("Degree distribution")
plt.xlabel("Degree")
plt.ylabel("Number of nodes")
plt.xticks(range(min(degrees) - 2, max(degrees) + 3))
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## Network Metrics

### Shortest Path Length

A path in a network is a sequence of links that connects two nodes. For instance,
$P_1 = 1 \to 3 \to 2 \to 4$ is a path between nodes $1$ and $4$ traversing along nodes $3$ and
$2$. The path length is the number of links traversed along the path, so $|P_1| = 3$ for this
path. The shortest path length between two nodes is the smallest length among all the paths
that connect them.

The shortest path lengths of all pairs of nodes form the matrix $\mathcal{D}$, which we can
compute from the powers of the adjacency matrix $A$. The entry $(A^{k})_{ij}$ is the number of
ways to go from node $i$ to node $j$ in exactly $k$ steps:

$$
\begin{aligned}
A_{ij} &= \text{the number of routes of length 1 between nodes } i \text{ and } j,\\
(A^{2})_{ij} &= \text{the number of routes of length 2 between nodes } i \text{ and } j,\\
&\ \ \vdots \\
(A^{n-1})_{ij} &= \text{the number of routes of length } n-1 \text{ between nodes } i \text{ and } j.
\end{aligned}
$$

Such a route may visit the same node more than once (it is called a walk), but the shortest one
never does. Hence, the shortest path length $\mathcal{D}_{ij}$ is the smallest $k$ for which
$(A^{k})_{ij} > 0$. In a network with $n$ nodes a shortest path has at most $n - 1$ links, so
it is enough to compute the powers up to $A^{n-1}$.

**Algorithm: shortest path lengths of all pairs from matrix powers**

*Input:* the adjacency matrix $A$ of an unweighted network with $n$ nodes.
*Output:* the matrix of shortest path lengths $\mathcal{D}$.

1. Set every entry of $\mathcal{D}$ to $\infty$, and the diagonal to $0$.
2. Set $A_{k} = I$, the $n \times n$ identity matrix.
3. For $k = 1, 2, \dots, n - 1$:
    - multiply $A_{k}$ by $A$, so that $A_{k} = A^{k}$;
    - for every pair $(i, j)$ with $\mathcal{D}_{ij} = \infty$ and $(A_{k})_{ij} > 0$, set
      $\mathcal{D}_{ij} = k$;
    - stop if no entry of $\mathcal{D}$ equals $\infty$, as all the distances have been found.
4. Return $\mathcal{D}$. An entry that is still $\infty$ belongs to a pair of nodes with no path
   between them.

This algorithm is easy to follow but slow for large networks; for faster ones, see the
[NetworkX documentation](https://networkx.org/documentation/stable/reference/algorithms/shortest_paths.html).

```python
n = A.shape[0]

D = np.full((n, n), np.inf)
np.fill_diagonal(D, 0)

A_k = np.eye(n, dtype=int)
for k in range(1, n):
    A_k = A_k @ A
    # Pairs not reached so far that have a route of length k
    reached_now = np.isinf(D) & (A_k > 0)
    D[reached_now] = k
    if not np.isinf(D).any():  # all the distances have been found
        break

print("Shortest path length matrix D =")
print(D.astype(int))
```

### Average Path Length and Diameter

The average shortest path length, $\bar{d}$, is the mean of the shortest path lengths across the
network,

$$
\bar{d} = \frac{1}{n (n-1)}\sum_{i \neq j} \mathcal{D}_{ij}.
$$

The diameter, $R$, of a network is the maximum of the shortest path lengths across all pairs of
nodes, namely the greatest entry of the matrix $\mathcal{D}$.

```python
average_path_length = D.sum() / (n * (n - 1))
diameter = int(D.max())

print(f"Average shortest path length = {average_path_length:.4f}")
print(f"Diameter = {diameter}")
```

### Clustering Coefficients

The clustering coefficient measures how often two neighbours of a node are themselves linked,
so that the three nodes form a triangle. There are two types of clustering coefficients in the
literature.

**(a) The global clustering coefficient**, also known as **transitivity**, considers the connected
triplets of nodes, where a triplet is a node together with two of its neighbours (a path of
length 2 centred at that node). It is the fraction of triplets that are closed by a third link.
Figure 3 shows the difference on our regular network.

```python
# Find one triangle and one open triplet (a node whose two neighbours are not linked)
triangle = next(sorted(c) for c in nx.enumerate_all_cliques(G) if len(c) == 3)
centre, u, v = next((c, u, v) for c in sorted(G) for u in sorted(G[c]) for v in sorted(G[c])
                    if u < v and not G.has_edge(u, v))

fig, axes = plt.subplots(1, 2, figsize=(10, 4.5))
panels = [
    (axes[0], [(triangle[0], triangle[1]), (triangle[1], triangle[2]), (triangle[0], triangle[2])], [],
     triangle, f"(a) A triangle: {triangle[0]}, {triangle[1]}, {triangle[2]}"),
    (axes[1], [(u, centre), (centre, v)], [(u, v)],
     [u, centre, v], f"(b) An open triplet: {u}-{centre}-{v}, centred at {centre}"),
]
for ax, highlighted, missing, nodes, title in panels:
    nx.draw_networkx_edges(G, pos, ax=ax, edge_color="lightgray", width=1.5)
    nx.draw_networkx_edges(G, pos, ax=ax, edgelist=highlighted, edge_color="crimson", width=4)
    nx.draw_networkx_edges(G, pos, ax=ax, edgelist=missing, edge_color="crimson", width=2, style="dashed")
    nx.draw_networkx_nodes(G, pos, ax=ax, node_color=["crimson" if node in nodes else "skyblue" for node in G],
                           node_size=600)
    nx.draw_networkx_labels(G, pos, ax=ax, font_weight="bold")
    ax.set_title(title)
    ax.set_aspect("equal")
    ax.axis("off")
plt.tight_layout()
plt.show()
```

**Figure 3:** (a) A triangle of the regular network: three nodes that are all linked to each
other. (b) An open triplet: a node (the centre) together with two of its neighbours that are not
linked to each other (dashed line). If the dashed link existed, the triplet would be closed and
the three nodes would form a triangle.

Every triangle closes three triplets, one centred at each of its nodes, hence

$$
T = \frac{3 \times \text{number of triangles}}{\text{number of triplets}},
$$

$$
\text{number of triangles} = \frac{1}{6}\,\mathrm{Tr}\big(A^{3}\big), \qquad
\text{number of triplets} = \sum_{i=1}^{n}\binom{d_i}{2} = \frac{1}{2}\sum_{i=1}^{n} d_i (d_i - 1),
$$

where $d_{i}$ is the degree of node $i$. The entry $(A^{3})_{ii}$ counts the closed routes of
length 3 that start and end at node $i$; each triangle is counted six times in the trace, as it
can start at any of its three nodes and be traversed in two directions.

**(b) The local clustering coefficient** considers the links among the neighbours of a node, and
was introduced in [[5](#ref-5)] to quantify the small-worldness of a network. For a node $i$ with degree
$d_{i} \ge 2$,

$$
C_{i} = \frac{2 e_{i}}{d_{i}(d_{i}-1)},
$$

where $e_{i}$ is the number of links among the neighbours of node $i$, in other words, the
number of triangles that include node $i$, $e_{i} = (A^{3})_{ii}/2$. For a node with fewer than two
neighbours we set $C_{i} = 0$. The mean local clustering coefficient is the average over all the
nodes,

$$
C = \frac{1}{n}\sum_{i=1}^{n} C_{i}.
$$

**Note:** the two formulas are similar, but they are not the same. The transitivity divides the
total number of closed triplets by the total number of triplets of the whole network at once,
whereas $C$ computes this ratio node by node and then takes the average. They coincide when all
nodes have the same degree, as in a regular network.

```python
A_3 = A @ A @ A
degree = A.sum(axis=1)

n_triangles = np.trace(A_3) / 6
n_triplets = np.sum(degree * (degree - 1)) / 2
transitivity = 3 * n_triangles / n_triplets

e = np.diag(A_3) / 2  # triangles through each node
local_clustering = np.zeros(n)
enough_neighbours = degree >= 2
local_clustering[enough_neighbours] = 2 * e[enough_neighbours] / (degree * (degree - 1))[enough_neighbours]

print("A^3 =")
print(A_3)
print(f"Number of triangles = {n_triangles:.0f}")
print(f"Number of triplets  = {n_triplets:.0f}")
print(f"Transitivity T = {transitivity:.4f}")
print("Local clustering coefficients C_i:", {i + 1: round(c, 4) for i, c in enumerate(local_clustering)})
print(f"Mean local clustering coefficient C = {local_clustering.mean():.4f}")
```

### The Same Metrics with NetworkX

NetworkX has a function for each of these metrics, and they agree with the values we computed
by hand. Note that `average_shortest_path_length` and `diameter` raise an error when the network
is not connected, because some distances are then infinite; the function
[`calculate_distance_metrics`](../api/network_metrics.md) of the library handles that case, and
we use it in the [next tutorial](02_network_types.ipynb).

```python
print(f"Average shortest path length: {nx.average_shortest_path_length(G):.4f}")
print(f"Diameter: {nx.diameter(G)}")
print(f"Number of triangles: {sum(nx.triangles(G).values()) // 3}")
print(f"Transitivity (global clustering): {nx.transitivity(G):.4f}")
print(f"Local clustering (node by node): { {node: round(c, 4) for node, c in nx.clustering(G).items()} }")
print(f"Mean local clustering coefficient: {nx.average_clustering(G):.4f}")
```

## Summary

- A graph is a set of nodes, a set of links and a rule that says which nodes each link joins.
  Euler's argument on the Königsberg bridges only needs the degrees of the nodes.
- The adjacency matrix $A$ stores a network; its row sums are the degrees, and its powers count
  the routes of a given length between two nodes.
- The matrix of shortest path lengths $\mathcal{D}$ gives the average path length (its mean)
  and the diameter (its greatest entry).
- The transitivity $T$ and the mean local clustering coefficient $C$ both measure how many
  triangles a network has, the first for the whole network at once and the second node by node.

In the [next tutorial](02_network_types.ipynb) we use these metrics to compare regular, random,
small-world and scale-free networks.
